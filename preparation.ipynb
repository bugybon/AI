```python
import numpy as np
```

```python
array = input().split(" ")
array = [arr.split(",") for arr in array]

seasonToInt = {"Spring": 1, "Summer": 2, "Autumn":3, "Winter":4}
for a in array:
    a[0] = float(a[0])
    a[1] = seasonToInt[a[1]]
    a[2] = float(a[2])
    a[3] = float(a[3])

print(array)
```

```python
def computeDistance(a, b):
    return np.sqrt(np.sum([(ai-bi) ** 2 for ai, bi in zip(a, b)]))
```

```python
def knn(training, testing, k):
    result = [''] * len(testing)
    for i in range(len(testing)):
        distances = [(computeDistance(testing[i], trainer[:-1]), trainer[-1]) for trainer in training]
        distances.sort()
        distances = distances[:k]
        labels, counts = np.unique([d[1] for d in distances], return_counts=True)
        result[i] = labels[np.argmax(counts)]

    return result



```

```python
training_index = np.random.choice(len(array), int(len(array)*0.8), replace=False)
training_index.sort()
testing_index = [i for i in range(len(array)) if i not in training_index]
# print(testing_index, "\n", testing_index[:4])
training_data = [array[i] for i in training_index]
testing_data = [array[i] for i in testing_index]
print(training_data, "\n", testing_data)
```

```python
matches = np.array(knn(training_data, testing_data, 5))

def transformer(array, type):
    return np.array([a == type if 1 else 0 for a in array])

def classificationMetric(results, truths):
    TP = 0
    TN = 0
    FP = 0
    FN = 0
    for i in range(len(results)):
        if results[i] == 1:
            if results[i] == truths[i]:
                TP += 1
            else:
                FN += 1
        else:
            if results[i] == truths[i]:
                TN += 1
            else:
                FP += 1
    return TP, FP, FN, TN
categories = np.unique(matches)
for c in categories:
    TP, FP, FN, TN = classificationMetric(transformer(matches,c),transformer([t[-1] for t in testing_data], c))
    print(c)
    print('ACC: ', (TP + TN)/len(matches))
    print('Recall: ', TP/(TP + FN))
    print('Precision: ', TP/(TP + FP))
```
